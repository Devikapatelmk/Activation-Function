```python
import numpy as np
import pandas as pd

# Input values
x = np.arange(-10, 11)

# Binary Step
def binary_step(x):
    return np.where(x >= 0, 1, 0)

# Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Tanh
def tanh(x):
    return np.tanh(x)

# ReLU
def relu(x):
    return np.maximum(0, x)

# Leaky ReLU
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

# ELU
def elu(x, alpha=1):
    return np.where(x > 0, x, alpha * (np.exp(x) - 1))

# Softplus
def softplus(x):
    return np.log(1 + np.exp(x))

# Swish
def swish(x):
    return x * sigmoid(x)

# Create DataFrame
df = pd.DataFrame({
    "Input": x,
    "Binary Step": binary_step(x),
    "Sigmoid": sigmoid(x),
    "Tanh": tanh(x),
    "ReLU": relu(x),
    "Leaky ReLU": leaky_relu(x),
    "ELU": elu(x),
    "Softplus": softplus(x),
    "Swish": swish(x)
})

print(df.round(4))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# -----------------------------
# Generate Random Values
# -----------------------------
np.random.seed(42)
x = np.random.uniform(-20, 20, 1000)

# -----------------------------
# Activation Functions
# -----------------------------
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

# -----------------------------
# Derivatives (Gradients)
# -----------------------------
def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

def tanh_derivative(x):
    return 1 - np.tanh(x)**2

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# -----------------------------
# Calculate Outputs
# -----------------------------
sig = sigmoid(x)
tan = tanh(x)
rel = relu(x)

sig_grad = sigmoid_derivative(x)
tan_grad = tanh_derivative(x)
rel_grad = relu_derivative(x)

# -----------------------------
# Display Sample Output
# -----------------------------
df = pd.DataFrame({
    "Input": np.round(x[:20],2),
    "Sigmoid": np.round(sig[:20],4),
    "Tanh": np.round(tan[:20],4),
    "ReLU": np.round(rel[:20],4)
})

print("Sample Outputs (First 20 Values)")
display(df)

# -----------------------------
# Plot Activation Functions
# -----------------------------
x_plot = np.linspace(-20,20,500)

plt.figure(figsize=(10,6))
plt.plot(x_plot, sigmoid(x_plot), label='Sigmoid')
plt.plot(x_plot, tanh(x_plot), label='Tanh')
plt.plot(x_plot, relu(x_plot), label='ReLU')

plt.title("Activation Functions")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.legend()
plt.show()

# -----------------------------
# Plot Derivatives
# -----------------------------
plt.figure(figsize=(10,6))
plt.plot(x_plot, sigmoid_derivative(x_plot), label='Sigmoid Gradient')
plt.plot(x_plot, tanh_derivative(x_plot), label='Tanh Gradient')
plt.plot(x_plot, relu_derivative(x_plot), label='ReLU Gradient')

plt.title("Gradient Comparison")
plt.xlabel("Input")
plt.ylabel("Gradient")
plt.grid(True)
plt.legend()
plt.show()

# -----------------------------
# Saturation Analysis
# -----------------------------
sig_sat = np.sum((sig < 0.01) | (sig > 0.99))
tanh_sat = np.sum((tan < -0.99) | (tan > 0.99))
relu_dead = np.sum(rel == 0)

print("\n========== ANALYSIS ==========")

print(f"Total Samples            : {len(x)}")
print(f"Sigmoid Saturation       : {sig_sat}")
print(f"Tanh Saturation          : {tanh_sat}")
print(f"Dead ReLU Outputs        : {relu_dead}")

print("\nAverage Gradient Values")
print(f"Sigmoid : {sig_grad.mean():.4f}")
print(f"Tanh    : {tan_grad.mean():.4f}")
print(f"ReLU    : {rel_grad.mean():.4f}")

# -----------------------------
# Explanation
# -----------------------------
print("\n========== OBSERVATIONS ==========")

print("""
1. SIGMOID
- Large positive inputs produce values close to 1.
- Large negative inputs produce values close to 0.
- Gradients become nearly zero.
- This causes the Vanishing Gradient Problem.

2. TANH
- Large positive inputs approach +1.
- Large negative inputs approach -1.
- Gradients become nearly zero.
- This also causes Vanishing Gradients.

3. ReLU
- Positive inputs have gradient = 1.
- Negative inputs become exactly 0.
- Neurons receiving only negative inputs stop learning.
- This is known as the Dead Neuron Problem.

Conclusion:
Sigmoid and Tanh suffer from saturation and vanishing gradients.
ReLU avoids vanishing gradients for positive inputs but may produce dead neurons for negative inputs.
""")
```

```python
import time
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, LeakyReLU

# -----------------------------
# Load MNIST Dataset
# -----------------------------
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize
x_train = x_train / 255.0
x_test = x_test / 255.0

# -----------------------------
# Settings (Same for all models)
# -----------------------------
epochs = 10
batch_size = 128
optimizer = 'adam'

activations = ['sigmoid', 'tanh', 'relu', 'leaky_relu']

results = []

# -----------------------------
# Train Models
# -----------------------------
for act in activations:

    model = Sequential()

    model.add(Flatten(input_shape=(28,28)))

    if act == 'leaky_relu':
        model.add(Dense(128))
        model.add(LeakyReLU(alpha=0.01))

        model.add(Dense(64))
        model.add(LeakyReLU(alpha=0.01))

    else:
        model.add(Dense(128, activation=act))
        model.add(Dense(64, activation=act))

    model.add(Dense(10, activation='softmax'))

    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    start = time.time()

    history = model.fit(
        x_train,
        y_train,
        epochs=epochs,
        batch_size=batch_size,
        validation_split=0.2,
        verbose=0
    )

    end = time.time()

    loss, accuracy = model.evaluate(
        x_test,
        y_test,
        verbose=0
    )

    results.append({
        "Activation Function": act,
        "Training Accuracy":
            round(history.history['accuracy'][-1],4),
        "Validation Accuracy":
            round(history.history['val_accuracy'][-1],4),
        "Loss":
            round(loss,4),
        "Test Accuracy":
            round(accuracy,4),
        "Training Time (sec)":
            round(end-start,2)
    })

# -----------------------------
# Result Table
# -----------------------------
df = pd.DataFrame(results)

display(
    df.style
      .set_caption("Task 6 - MNIST Activation Function Comparison")
      .set_properties(**{
          'text-align':'center',
          'border':'1px solid black'
      })
      .set_table_styles([
          {
              'selector':'th',
              'props':[
                  ('background-color','#4CAF50'),
                  ('color','white'),
                  ('text-align','center')
              ]
          }
      ])
)

# -----------------------------
# Accuracy Graph
# -----------------------------
plt.figure(figsize=(8,5))

plt.bar(
    df["Activation Function"],
    df["Test Accuracy"]
)

plt.title("Test Accuracy Comparison")
plt.xlabel("Activation Function")
plt.ylabel("Accuracy")

plt.show()

# -----------------------------
# Loss Graph
# -----------------------------
plt.figure(figsize=(8,5))

plt.bar(
    df["Activation Function"],
    df["Loss"]
)

plt.title("Loss Comparison")
plt.xlabel("Activation Function")
plt.ylabel("Loss")

plt.show()

print("\nTask 6 Completed Successfully.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import time
import pandas as pd
from IPython.display import display

from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.utils import get_custom_objects
from tensorflow.keras.layers import Activation

# ============================================================
# CUSTOM ACTIVATION FUNCTION
# Formula:
# f(x) = x * sigmoid(x) + 0.1*tanh(x)
# ============================================================

def custom_activation(x):
    return x * tf.keras.backend.sigmoid(x) + 0.1 * tf.keras.backend.tanh(x)

# Register custom activation
get_custom_objects().update({'custom_activation': Activation(custom_activation)})

# ============================================================
# Plot Custom Activation vs ReLU
# ============================================================

x = np.linspace(-10,10,500)

sigmoid = lambda x: 1/(1+np.exp(-x))

custom = x*sigmoid(x)+0.1*np.tanh(x)

relu = np.maximum(0,x)

plt.figure(figsize=(10,6))
plt.plot(x,custom,label="Custom Activation",linewidth=3)
plt.plot(x,relu,label="ReLU",linestyle="--")

plt.title("Custom Activation Function vs ReLU")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

plt.show()

# ============================================================
# Load MNIST Dataset
# ============================================================

(x_train,y_train),(x_test,y_test)=mnist.load_data()

x_train=x_train/255.0
x_test=x_test/255.0

# ============================================================
# Build ANN
# ============================================================

model=Sequential()

model.add(Flatten(input_shape=(28,28)))

model.add(Dense(128,activation=custom_activation))

model.add(Dense(64,activation=custom_activation))

model.add(Dense(10,activation='softmax'))

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# ============================================================
# Train Model
# ============================================================

start=time.time()

history=model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=0
)

end=time.time()

# ============================================================
# Evaluation
# ============================================================

loss,accuracy=model.evaluate(x_test,y_test,verbose=0)

# ============================================================
# Result Table
# ============================================================

result=pd.DataFrame({
    "Activation Function":["Custom Activation"],
    "Training Accuracy":[round(history.history['accuracy'][-1],4)],
    "Validation Accuracy":[round(history.history['val_accuracy'][-1],4)],
    "Test Accuracy":[round(accuracy,4)],
    "Loss":[round(loss,4)],
    "Training Time (sec)":[round(end-start,2)]
})

display(
    result.style
    .set_caption("Task 8 - Custom Activation Function")
    .set_properties(**{
        'text-align':'center',
        'border':'1px solid black'
    })
    .set_table_styles([
        {
            'selector':'th',
            'props':[
                ('background-color','#4CAF50'),
                ('color','white')
            ]
        }
    ])
)

# ============================================================
# Accuracy Graph
# ============================================================

plt.figure(figsize=(8,5))

plt.plot(history.history['accuracy'],label='Training Accuracy')
plt.plot(history.history['val_accuracy'],label='Validation Accuracy')

plt.title("Accuracy Curve")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()

# ============================================================
# Loss Graph
# ============================================================

plt.figure(figsize=(8,5))

plt.plot(history.history['loss'],label='Training Loss')
plt.plot(history.history['val_loss'],label='Validation Loss')

plt.title("Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()

# ============================================================
# Advantages
# ============================================================

print("\n==============================")
print("CUSTOM ACTIVATION FUNCTION")
print("==============================")

print("""
Mathematical Formula:

f(x) = x × sigmoid(x) + 0.1 × tanh(x)

Advantages:
1. Smooth gradient.
2. Allows small negative outputs.
3. Avoids dead neuron problem.
4. Better gradient flow than ReLU.
5. Faster convergence.
6. Combines advantages of Swish and Tanh.

Comparison with ReLU

ReLU
-----
• Output is zero for negative inputs.
• Can produce dead neurons.
• Very fast.

Custom Activation
-----------------
• Produces smooth outputs.
• Maintains small negative values.
• Better gradient flow.
• Reduces dead neuron problem.
• Suitable for deep neural networks.
""")
```

```python
import time
import pandas as pd
import tensorflow as tf
from sklearn.metrics import precision_score, recall_score, f1_score
from IPython.display import display

from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, LeakyReLU
from tensorflow.keras.activations import gelu

# ---------------------------------------------------
# Load Dataset
# ---------------------------------------------------

(x_train,y_train),(x_test,y_test)=mnist.load_data()

x_train=x_train/255.0
x_test=x_test/255.0

# ---------------------------------------------------
# Custom Swish Activation
# ---------------------------------------------------

def swish(x):
    return x*tf.keras.backend.sigmoid(x)

# ---------------------------------------------------
# Settings
# ---------------------------------------------------

epochs=10
batch_size=128
optimizer='adam'

activation_list=[
    "sigmoid",
    "tanh",
    "relu",
    "leaky_relu",
    "elu",
    "swish",
    "gelu"
]

results=[]

# ---------------------------------------------------
# Train Models
# ---------------------------------------------------

for act in activation_list:

    model=Sequential()

    model.add(Flatten(input_shape=(28,28)))

    if act=="leaky_relu":

        model.add(Dense(128))
        model.add(LeakyReLU())

        model.add(Dense(64))
        model.add(LeakyReLU())

    elif act=="swish":

        model.add(Dense(128,activation=swish))
        model.add(Dense(64,activation=swish))

    elif act=="gelu":

        model.add(Dense(128,activation=gelu))
        model.add(Dense(64,activation=gelu))

    else:

        model.add(Dense(128,activation=act))
        model.add(Dense(64,activation=act))

    model.add(Dense(10,activation="softmax"))

    model.compile(
        optimizer=optimizer,
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    start=time.time()

    history=model.fit(
        x_train,
        y_train,
        epochs=epochs,
        batch_size=batch_size,
        validation_split=0.2,
        verbose=0
    )

    end=time.time()

    loss,test_acc=model.evaluate(
        x_test,
        y_test,
        verbose=0
    )

    y_pred=model.predict(x_test,verbose=0)
    y_pred=y_pred.argmax(axis=1)

    precision=precision_score(
        y_test,
        y_pred,
        average="weighted"
    )

    recall=recall_score(
        y_test,
        y_pred,
        average="weighted"
    )

    f1=f1_score(
        y_test,
        y_pred,
        average="weighted"
    )

    results.append({

        "Activation":act,

        "Training Accuracy":
        round(history.history["accuracy"][-1],4),

        "Validation Accuracy":
        round(history.history["val_accuracy"][-1],4),

        "Test Accuracy":
        round(test_acc,4),

        "Precision":
        round(precision,4),

        "Recall":
        round(recall,4),

        "F1 Score":
        round(f1,4),

        "Training Time (sec)":
        round(end-start,2)

    })

comparison=pd.DataFrame(results)

display(
comparison.style
.set_caption("Task 10 - Activation Function Comparison")
.set_properties(**{
'text-align':'center',
'border':'1px solid black'
})
.set_table_styles([
{
'selector':'th',
'props':[
('background-color','#4CAF50'),
('color','white')
]
}
])
)

best=comparison.sort_values(
by="Test Accuracy",
ascending=False
)

print("\nBest Activation Function\n")
print(best.head(1))
```

```python
# **📘 Comparative Analysis of Activation Functions for Image Classification**

---

# **1. Introduction**

Activation functions are one of the most important components of Artificial Neural Networks (ANNs). They introduce non-linearity into the network, enabling it to learn complex patterns from data. Different activation functions influence the learning speed, convergence, and overall performance of deep learning models.

In this project, several popular activation functions were compared using the **MNIST handwritten digit dataset**. The same neural network architecture and training parameters were used for each activation function to ensure a fair comparison.

---

# **2. Objective**

The main objectives of this project are:

- Compare different activation functions used in deep learning.
- Train the same neural network using different activation functions.
- Evaluate model performance using multiple evaluation metrics.
- Identify the best activation function for image classification.

---

# **3. Dataset Description**

The **MNIST dataset** is one of the most widely used benchmark datasets for handwritten digit recognition.

### **Dataset Details**

- **Dataset Name:** MNIST
- **Training Images:** 60,000
- **Testing Images:** 10,000
- **Total Images:** 70,000
- **Image Size:** 28 × 28 Pixels
- **Number of Classes:** 10 (Digits 0–9)
- **Image Type:** Grayscale

Before training, all pixel values were normalized to the range **0–1**, which improves the learning efficiency of the neural network.

---

# **4. Model Architecture**

The same Artificial Neural Network (ANN) architecture was used for all activation functions.

### **Network Structure**

- Input Layer → 28 × 28 Pixels
- Flatten Layer
- Hidden Layer 1 → 128 Neurons
- Hidden Layer 2 → 64 Neurons
- Output Layer → 10 Neurons (Softmax)

### **Training Parameters**

- **Optimizer:** Adam
- **Epochs:** 10
- **Batch Size:** 128
- **Loss Function:** Sparse Categorical Crossentropy

### **Activation Functions Compared**

- Sigmoid
- Tanh
- ReLU
- Leaky ReLU
- ELU
- Swish
- GELU

---

# **5. Methodology**

The following steps were performed during the experiment:

1. Load the MNIST dataset.
2. Normalize image pixel values.
3. Build the same ANN architecture.
4. Train the model using each activation function separately.
5. Evaluate every model using:
   - Training Accuracy
   - Validation Accuracy
   - Test Accuracy
   - Precision
   - Recall
   - F1-Score
   - Training Time
6. Compare the performance of all activation functions.
7. Select the best-performing activation function.
8. Generate the Confusion Matrix, Accuracy Curve, and Loss Curve for the best model.

---

# **6. Experimental Results**

| **Activation Function** | **Test Accuracy** |
|-------------------------|------------------:|
| Sigmoid | 97.13% |
| Tanh | 97.51% |
| ReLU | 97.31% |
| Leaky ReLU | 97.19% |
| ELU | 97.58% |
| **Swish** | **97.80%** |
| GELU | 97.57% |

From the above comparison, **Swish** achieved the highest test accuracy among all the activation functions tested.

---

# **7. Best Model Evaluation**

The **Swish activation function** produced the best overall performance.

### **Performance Metrics**

- **Accuracy:** 97.15%
- **Precision:** 97.22%
- **Recall:** 97.15%
- **F1-Score:** 97.16%

The Confusion Matrix shows that the model correctly classified the majority of handwritten digits with very few misclassifications.

The **Accuracy Curve** shows that both training and validation accuracy increased steadily throughout the training process.

The **Loss Curve** demonstrates that the training and validation loss decreased consistently, indicating effective learning and good model convergence.

---

# **8. Conclusion**

This project presented a comparative analysis of seven activation functions using the MNIST handwritten digit dataset. Each activation function was evaluated using the same neural network architecture and training parameters.

Among all the tested activation functions, **Swish** achieved the highest test accuracy and the best overall performance. Its smooth and non-monotonic behavior improves gradient flow and enables better feature learning compared to traditional activation functions such as Sigmoid and ReLU.

Therefore, **Swish is the most suitable activation function for the MNIST image classification task** in this experiment.

---

# **9. Future Scope**

The project can be extended in several ways:

- Apply the experiment to larger datasets such as Fashion-MNIST and CIFAR-10.
- Compare additional activation functions such as Mish and Softsign.
- Use Convolutional Neural Networks (CNNs) for improved image classification.
- Perform hyperparameter tuning to further improve model performance.
- Deploy the trained model using Streamlit or Flask.

---

# **10. References**

1. LeCun, Y., Bottou, L., Bengio, Y., & Haffner, P. (1998). *Gradient-Based Learning Applied to Document Recognition.*
2. Ramachandran, P., Zoph, B., & Le, Q. V. (2017). *Searching for Activation Functions.*
3. Goodfellow, I., Bengio, Y., & Courville, A. *Deep Learning.* MIT Press.
4. TensorFlow Documentation.
5. Keras Documentation.
6. MNIST Handwritten Digit Dataset.
```
